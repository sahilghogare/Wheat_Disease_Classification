## Define Human Presence Detection Function

### Subtask:
Create a Python function that takes an image path, loads the image, and uses a pre-trained classifier (e.g., Haar cascade for face detection) to detect human presence (e.g., faces). This function will return a boolean indicating if human presence is detected.


**Reasoning**:
To define the `has_human_presence` function as specified, I need to import the `cv2` library, load the Haar cascade classifier for face detection, and implement the image processing steps to detect faces and return a boolean.



```python
import cv2

def has_human_presence(image_path):
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

    image = cv2.imread(image_path)
    if image is None:
        print(f"Warning: Could not read image {image_path}")
        return False

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    return len(faces) > 0
```

```python
import cv2
import matplotlib.pyplot as plt

# -----------------------------
# Function to crop bottom % of image
# -----------------------------
def crop_bottom_percentage(image_path, crop_percent=20):
    """
    image_path: path to your test image
    crop_percent: how much percentage to crop from bottom (0–100)
    """
    img = cv2.imread(image_path)

    if img is None:
        print("Error: Image not found!")
        return

    h, w, c = img.shape
    crop_pixels = int((crop_percent / 100) * h)

    # Crop from top to (height - crop_pixels)
    cropped_img = img[:h - crop_pixels, :, :]

    print(f"Original height: {h} px")
    print(f"Cropped pixels: {crop_pixels} px ({crop_percent}% of image height)")
    print(f"New height: {cropped_img.shape[0]} px")

    # Convert BGR → RGB for display
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    cropped_rgb = cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB)

    # Show images
    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.title("Original Image")
    plt.imshow(img_rgb)
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.title(f"Cropped Image ({crop_percent}%)")
    plt.imshow(cropped_rgb)
    plt.axis("off")

    plt.show()

    # Save cropped image
    output_path = "cropped_result.jpg"
    cv2.imwrite(output_path, cropped_img)
    print(f"Cropped image saved as: {output_path}")

# --------------------------------
# TEST THE FUNCTION ON YOUR IMAGE
# --------------------------------
test_image = "/content/drive/MyDrive/Wheat_Images/Leaf Rust/06921.jpg"  # <-- change this to your file path
crop_bottom_percentage(test_image, crop_percent=15)
```

```python
reader = easyocr.Reader(['en'])   # English OCR

def contains_text(image_path, min_text_len=1):
    try:
        result = reader.readtext(image_path, detail=0)
        extracted_text = " ".join(result)

        # If text length is more than 5 characters → treat as text image
        if len(extracted_text.strip()) >= min_text_len:
            return True
        return False

    except Exception as e:
        print("OCR error:", e)
        return False
```

```python
import cv2
import os
import shutil

def contains_unwanted_objects(img_path):

    results = model(img_path)[0]   # YOLO inference

    for box in results.boxes:
        cls = int(box.cls[0])
        label = results.names[cls]

        if label in UNWANTED_LABELS:
            print(f"Detected: {label} in {img_path}")
            return True

    return False
```

```python
def move_safely(src_path, dst_folder):
    filename = os.path.basename(src_path)
    base, ext = os.path.splitext(filename)

    counter = 1
    new_name = filename

    while os.path.exists(os.path.join(dst_folder, new_name)):
        new_name = f"{base}_{counter}{ext}"
        counter += 1

    shutil.move(src_path, os.path.join(dst_folder, new_name))
```

```python
import os
import shutil
import pytesseract
from PIL import Image


# ---- PATHS ----
folder_path = "/content/drive/MyDrive/Wheat_Images/Healthy Wheat"      # main image folder
junk_path = "/content/drive/MyDrive/Wheat_Images/Junk"    # junk folder

# create junk folder if it does not exist
os.makedirs(junk_path, exist_ok=True)

# supported image formats
valid_ext = (".jpg", ".jpeg", ".png")

# ---- PROCESS IMAGES ----
moved = 0

for filename in os.listdir(folder_path):
    if filename.lower().endswith(valid_ext):
        img_path = os.path.join(folder_path, filename)

        try:
            img = Image.open(img_path)

            # extract text using OCR
            text = pytesseract.image_to_string(img)

            # if any detectable text exists
            if text.strip() != "":
                print(f"[TEXT FOUND] {filename} → moving to junk folder")

                shutil.move(img_path, os.path.join(junk_path, filename))
                moved += 1

        except Exception as e:
            print(f"Error processing {filename}: {e}")

print(f"\n✔ Done! Moved {moved} images to junk folder.")
```

```python
import os
import shutil
from ultralytics import YOLO
from PIL import Image

# Load pretrained YOLOv8 model
model = YOLO("yolov8n.pt")   # small and fast; good enough for filtering

# Paths
folder_path = "/content/drive/MyDrive/Wheat_Images/Healthy Wheat"      # main image folder
junk_path = "/content/drive/MyDrive/Wheat_Images/Junk"    # junk folder

# Create junk folder
os.makedirs(junk_path, exist_ok=True)

# Unwanted classes to detect
UNWANTED = {
    "person", "hand", "finger", "arm",
    "car", "truck", "tractor", "bus", "motorcycle",
    "bicycle", "machinery",
    "building"
}

moved = 0

# Go through all images
for fname in os.listdir(folder_path):
    if not fname.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    img_path = os.path.join(folder_path, fname)

    # run YOLO detection
    results = model(img_path, verbose=False)

    unwanted_found = False

    # Check detected objects
    for box in results[0].boxes:
        cls = int(box.cls[0])
        cls_name = model.names[cls]

        if cls_name in UNWANTED:
            unwanted_found = True
            break

    # Move if unwanted object detected
    if unwanted_found:
        print(f"[UNWANTED: {cls_name}]  → Moving {fname} to junk folder")
        shutil.move(img_path, os.path.join(junk_path, fname))
        moved += 1

print(f"\n✔ Done! Moved {moved} images to junk folder.")
```

```python


import os
from PIL import Image
import imagehash
from collections import defaultdict

# Path to your folder in Google Drive
folder_path = "/content/drive/MyDrive/Wheat_Images/Junk"  # change to your folder

# Supported image formats
image_extensions = (".jpg", ".jpeg", ".png")

# Step 1: Compute hashes for all images
hash_dict = defaultdict(list)
for filename in os.listdir(folder_path):
    if filename.lower().endswith(image_extensions):
        filepath = os.path.join(folder_path, filename)
        try:
            img = Image.open(filepath)
            img_hash = imagehash.phash(img)  # perceptual hash
            hash_dict[str(img_hash)].append(filename)
        except Exception as e:
            print(f"Error processing {filename}: {e}")

# Step 2: Preview duplicates
print("Potential duplicates:")
for files in hash_dict.values():
    if len(files) > 1:
        print(files)

# Step 3: Rename duplicates
for files in hash_dict.values():
    if len(files) > 1:
        original = files[0]
        for idx, dup in enumerate(files[1:], start=1):
            ext = os.path.splitext(dup)[1]
            new_name = f"{os.path.splitext(original)[0]}_{idx}{ext}"
            os.rename(os.path.join(folder_path, dup), os.path.join(folder_path, new_name))

print("Duplicates renamed! Files are now sorted alphabetically.")
```
